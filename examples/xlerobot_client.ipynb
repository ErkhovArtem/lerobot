```python
import time
import numpy as np

from lerobot.utils.rotation import Rotation
from lerobot.robots.xlerobot import XLerobotClient, XLerobotClientConfig
```

```python
ROBOT_IP = "0.0.0.0"

robot_config = XLerobotClientConfig(remote_ip=ROBOT_IP, id="my_xlerobot")
robot = XLerobotClient(robot_config)

try:
    robot.connect()
    print("Подключено к роботу")
except Exception as e:
    print(f"Ошибка: {e}")
```

```python
obs = robot.get_observation()
print(obs)
```

```python
delta_eef = {
    "enabled": True,
    "target_x": 0.0, 
    "target_y": 0.0, 
    "target_z": 0., 
    "target_wx": 0., 
    "target_wy": 0., 
    "target_wz": 0., 
    "gripper_vel": 0., 
}
```

```python
robot.calculate_action(delta_eef)
```

```python
# del obs['observation.state']
obs['shoulder_pan.pos'] = -20.
obs
```

```python
robot.send_action(obs)
```

```python
robot.disconnect()
```
